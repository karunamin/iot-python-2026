```python
import cv2
import time
import csv
from ultralytics import YOLO

model = YOLO('yolo11n.pt')

cap = cv2.VideoCapture('./sample02.mp4')

log_file = open('person_count_log.csv', 'w', newline='')
writer = csv.writer(log_file)
writer.writerow(['time', 'person_count'])

prev_time = time.time()

while True:
    ret, frame = cap.read()
    if not ret:
        break

    results = model(frame, conf=0.5)

    person_count = 0

    for box in results[0].boxes:
        cls_id = int(box.cls[0])
        label = model.names[cls_id]

        if label == 'person':
            person_count += 1

    frame = results[0].plot()

    cv2.putText(
    frame,
    f'People: {person_count}',
    (20, 50),
    cv2.FONT_HERSHEY_SIMPLEX,
    1,
    (0, 255, 0),
    2
    )

    now = time.time()
    if now - prev_time >= 1:
        current_time = time.strftime('%Y-%m-%d %H:%M:%S')
        writer.writerow([current_time, person_count])
        prev_time = now


    cv2.imshow('people', frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
log_file.close()
cv2.destroyAllWindows()
```
